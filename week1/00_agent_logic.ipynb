```python
import random
import time
target = 10
```

```python
loop_count=0
insan_girdisi=random.randint(11,20)
while True:
    insan_girdisi = int(input(f"Başlangıç: {insan_girdisi}; hedef: {target} "))
    loop_count += 1
    if insan_girdisi == target:
        print(f"{loop_count} model çağrısında göre tamamlandı.")
        break

    if loop_count > 10:
        print("Maksimum model çağrısına ulaşıldı. Agent görevi tamamlamadan sonlandırıldı.")
        break

    time.sleep(0.5)
    
```
